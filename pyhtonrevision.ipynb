```python
from logging import exception
import random
target_guess = random.randint(1,100)

game_menu = input(" enter 1 for start : ")
if game_menu == "1":
  game = True
else:
  game = False

number_of_try = 0
number_of_try_left = 10

while(game == True):
  try:
    user_guess = int(input("enter your guess between 1 to 100: "))
    number_of_try +=1
    number_of_try_left = 10 -number_of_try
    if number_of_try_left > 0:
      if user_guess == target_guess:
        print(f"you win and number of try is {number_of_try}")
        game = False
      elif user_guess < target_guess:
        print(f"please enter greater number --- chance available {number_of_try_left}")
      elif user_guess > target_guess:
        print(f"please enter smaller number  --- chance available {number_of_try_left}")
    else:
      print("you loose ")
      game = False
  except Exception as e:
      print(e)
```

```python
def add(a, b):
    return a + b

def sub(a, b):
    return a - b

def mul(a, b):
    return a * b

def div(a, b):
    return a / b


out = float(input("Enter number: "))

while True:
    operation = input("+ add   - sub   * mul   / div   = result   q = quit: ")

    if operation == "q":
        break

    if operation == "=":
        print("Result:", out)
        continue

    inp = float(input("Enter number: "))

    if operation == "+":
        out = add(out, inp)
    elif operation == "-":
        out = sub(out, inp)
    elif operation == "*":
        out = mul(out, inp)
    elif operation == "/":
        if inp == 0:
            print("Cannot divide by zero")
            continue
        out = div(out, inp)
    else:
        print("Invalid operation")
        continue

    print("Result:", out)
```

```python
# function to reverse a list
def reverse(lis):
  for i in range(len(lis)//2):
    lis[i],lis[len(lis)-i-1] = lis[len(lis)-i-1],lis[i]
  return lis
```

```python
# Flatten a list with n levels of nesting without recursion
def flat(input_list):
  # Create a mutable copy of the input list to process iteratively
  temp_list = list(input_list)

  # Continue as long as there is at least one nested list in temp_list
  while any(isinstance(item, list) for item in temp_list):
    new_temp_list = []
    for item in temp_list:
      if isinstance(item, list):
        # If an item is a list, extend new_temp_list with its contents
        new_temp_list.extend(item)
      else:
        # Otherwise, append the non-list item directly
        new_temp_list.append(item)
    temp_list = new_temp_list # Update temp_list for the next iteration
  return temp_list
```

```python

def run_length_encoding(s):
    result = ""
    count = 1

    for i in range(1, len(s)):
        if s[i] == s[i - 1]:
            count += 1
        else:
            result += s[i - 1] + str(count)
            count = 1

    # Add the last character
    if len(s) > 0:
        result += s[-1] + str(count)

    return result


# Ex
s = "AAABBBCCDAA"
print(run_length_encoding(s))

```

```python
# Q38. Create a function that removes all HTML tags from a string.
def remove_html_tags(s):
  s = s.split()
  sent = ""
  for i in s:
    if not (i.endswith(">") and i.startswith("<")):
      sent = sent + " " + i
  return sent.strip()
```

```python
# pig latin
# If a word starts with a consonant, move the first consonant to the end and add "ay".
# If a word starts with a vowel, usually add "way".
def pig_latin(word):
  if word[0].lower() in "aeiou":
    return word + "way"
  else:
    return word[1:] + word[0] + "ay"
```

```python
pig_latin("pple")
```

```python
# Q49. Write a script to clean a text file by removing extra spaces and blank lines.
with open("file.txt") as f:
  data = f.readlines()
new_data = []
for line in data:
  line = " ".join(line.split())
  if line:
    new_data.append(line)
  data = "\n".join(new_data)
with open("file.txt", "w") as f:
  f.write(data)
```
