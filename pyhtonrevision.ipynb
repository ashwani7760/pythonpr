```python
from logging import exception
import random
target_guess = random.randint(1,100)

game_menu = input(" enter 1 for start : ")
if game_menu == "1":
  game = True
else:
  game = False

number_of_try = 0
number_of_try_left = 10

while(game == True):
  try:
    user_guess = int(input("enter your guess between 1 to 100: "))
    number_of_try +=1
    number_of_try_left = 10 -number_of_try
    if number_of_try_left > 0:
      if user_guess == target_guess:
        print(f"you win and number of try is {number_of_try}")
        game = False
      elif user_guess < target_guess:
        print(f"please enter greater number --- chance available {number_of_try_left}")
      elif user_guess > target_guess:
        print(f"please enter smaller number  --- chance available {number_of_try_left}")
    else:
      print("you loose ")
      game = False
  except Exception as e:
      print(e)
```

```python
# Q15. Create a program to print the Fibonacci series up to N terms.
fibonaci = [0,1]
n  = int(input("enter n "))
if n<=1:
    print(fibonaci)
else:
    for j in range(2,n):
      fibonaci.append(fibonaci[j-2]+ fibonaci[j-1])
print(fibonaci)
```

```python
#gcd

numbers = list(map(int, input("Enter numbers separated by space: ").split()))
gcd =1
n1 = numbers[0]
n2 = numbers[1]
minn= min(n1,n2)
for i in range(1,minn+1):
 if n1%i ==0 and n2%i==0:
  gcd = i
print(gcd)
```

```python
#lcm
numbers = list(map(int, input("Enter numbers separated by space: ").split()))
lcm =1
n1 = numbers[0]
n2 = numbers[1]
maxx= max(n1,n2)
while True :
  if maxx%n1 ==0 and maxx%n2==0:
    lcm = maxx
    break
  maxx+=1
print(lcm)
```

```python
#frequency of digit in integers
num = input("input: ")
freq = {}
for i in num:
  if i not in freq:
    freq[i] = 0
  freq[i] += 1
print(freq)
```

```python
def add(a, b):
    return a + b

def sub(a, b):
    return a - b

def mul(a, b):
    return a * b

def div(a, b):
    return a / b


out = float(input("Enter number: "))

while True:
    operation = input("+ add   - sub   * mul   / div   = result   q = quit: ")

    if operation == "q":
        break

    if operation == "=":
        print("Result:", out)
        continue

    inp = float(input("Enter number: "))

    if operation == "+":
        out = add(out, inp)
    elif operation == "-":
        out = sub(out, inp)
    elif operation == "*":
        out = mul(out, inp)
    elif operation == "/":
        if inp == 0:
            print("Cannot divide by zero")
            continue
        out = div(out, inp)
    else:
        print("Invalid operation")
        continue

    print("Result:", out)
```

```python
cal = input("enter").split()
op , nu  =[], []
for i in cal:
  if i in ["+","-","*","/"]:
    op.append(i)
  else:
    nu.append(int(i))

out = nu[0]
for i in range(len(op)):
  if op[i] == "+":
    out = out + nu[i+1]
  elif op[i] == "-":
    out = out - nu[i+1]
  elif op[i] =="*":
    out = out * nu[i+1]
  elif op[i] == "/":
    out = out / nu[i+1]
print(out)
```

```python
# function to reverse a list
def reverse(lis):
  for i in range(len(lis)//2):
    lis[i],lis[len(lis)-i-1] = lis[len(lis)-i-1],lis[i]
  return lis
```
